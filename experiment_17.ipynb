```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    'Days': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri'],
    'Study_Hours': [2, 3, 2.5, 4, 3.5],
    'Marks': [50, 60, 55, 70, 65],
    'Attendance': [60, 70, 65, 80, 75],
    'Sleep_Hours': [6, 7, 6, 8, 7],
    'Assignments_Completed': [1, 2, 1, 3, 2]
}
df = pd.DataFrame(data)
print(df)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Days': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri'],
    'Study_Hours': [2, 3, 2.5, 4, 3.5],
    'Marks': [50, 60, 55, 70, 65],
    'Attendance': [60, 70, 65, 80, 75],
    'Sleep_Hours': [6, 7, 6, 8, 7],
    'Assignments_Completed': [1, 2, 1, 3, 2]
}
df = pd.DataFrame(data)

bars = plt.bar(df['Days'], df['Marks'], color='red')
for bar in bars:
  y = bar.get_height()
  plt.text(bar.get_x() + bar.get_width() / 2, y, str(y),
           ha= 'center', va='bottom')
plt.title("aayush")
plt.xlabel("Days")
plt.ylabel("Marks")

plt.grid(axis='y')      #adds horizontal grid for readability
plt.show()
```

```python
#Conditional scatter plot
import matplotlib.pyplot as plt
import pandas as pd

# Assuming df is available from previous cells, we'll redefine it to ensure 'Result' can be added.
# Re-create df with original data to ensure consistency and then add the 'Result' column
data = {
    'Days': ['Mon', 'Tue', 'Wed', 'Thu', 'Fri'],
    'Study_Hours': [2, 3, 2.5, 4, 3.5],
    'Marks': [50, 60, 55, 70, 65],
    'Attendance': [60, 70, 65, 80, 75],
    'Sleep_Hours': [6, 7, 6, 8, 7],
    'Assignments_Completed': [1, 2, 1, 3, 2]
}
df = pd.DataFrame(data)

# Correctly assign the 'Result' column to the DataFrame
df['Result'] = ['Fail', 'Pass', 'Fail', 'Pass', 'Pass']

# Correct the syntax for defining 'colors' and the comparison operator
colors = ['red' if r == 'Fail' else 'green' for r in df['Result']]

plt.scatter(df['Study_Hours'], df['Marks'], c=colors)

plt.title("aayush")

plt.xlabel("Study Hours")

plt.ylabel("Marks")

plt.show()
```
